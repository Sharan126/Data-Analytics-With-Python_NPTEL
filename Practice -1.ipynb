```python
#Line Plot: Life Expectancy trend for one country
import matplotlib.pyplot as plt

india = df[df["country"] == "India"]

plt.plot(india["year"], india["lifeExp"])
plt.title("India: Life Expectancy Over Years")
plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.show()
```

```python
#Bar Plot: Average life expectancy by continent
import matplotlib.pyplot as plt

continent_life = df.groupby("continent")["lifeExp"].mean()

continent_life.plot(kind="bar")
plt.title("Average Life Expectancy by Continent")
plt.xlabel("Continent")
plt.ylabel("Avg Life Expectancy")
plt.show()
```

```python
#ractice 1: Load dataset and print basic details:Task,Load data,Print first 5 rows,Print shape
import pandas as pd

url = "https://raw.githubusercontent.com/jennybc/gapminder/main/inst/extdata/gapminder.tsv"
df = pd.read_csv(url, sep="\t")

print(df.head())
print(df.shape)
```

```python
#Practice 2: Find unique continents
df["continent"].unique()
```

```python
#Practice 3: Print all column names + data types
print(df.columns)
print(df.dtypes)
```

```python
#Practice 4: Filter only Asian countries
asia = df[df["continent"] == "Asia"]
asia.head()
```

```python
#Practice 5: Select data for India after year 1980
india_1980 = df[(df["country"] == "India") & (df["year"] > 1980)]
india_1980
```

```python
#Practice 6: Country with highest life expectancy in 2007
df_2007 = df[df["year"] == 2007]
df_2007.sort_values("lifeExp", ascending=False).head(1)
```

```python
#Practice 7: Plot GDP per capita vs Life Expectancy (Scatter)
import matplotlib.pyplot as plt

df_2007 = df[df["year"] == 2007]

plt.scatter(df_2007["gdpPercap"], df_2007["lifeExp"])
plt.title("GDP per Capita vs Life Expectancy (2007)")
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy")
plt.show()
```

```python
#Practice 8 (Important): Correlation between GDP and LifeExp
df_2007[["gdpPercap", "lifeExp"]].corr()
```
